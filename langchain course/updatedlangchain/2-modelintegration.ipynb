## STREAMING AND BATCH



### STREAMING

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, partivularly for longer responses. Calling Stream() returns an iterator that yeilds output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
model.invoke("write me a 200 words paragraph on AI")
```

```python
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
for chunk in model.stream("write me a 200 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

```python
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
for chunk in model.stream("why is sky blue"):
    print(chunk.text, end="|", flush=True)
```

### BATCH

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
responses = model.batch([
    "why do parrots talk?",
    "how do birds fly?",
    "what is quantum computing?"
])

for response in responses:
    print(response.content)
```

```python
responses = model.batch([
    "why do parrots talk?",
    "how do birds fly?",
    "what is quantum computing?",
    "who are u?",
    "how do babies learn to talk?",
    "worlds hardest job?"
    ],
    config={
        'max_concurrency':5, #limits to 5 parallel calls
    }
)

for response in responses:
    print(response.content)
```
